# Kimia Farma Performance Analysis 2020–2023

```python
import pandas as pd
import numpy as np
```

```python
df_transaction = pd.read_csv('/kimia-farma-dashboard/raw/kf_final_transaction.csv')
df_transaction.head()
```

```python
df_branch = pd.read_csv('/kimia-farma-dashboard/raw/kf_kantor_cabang.csv')
df_branch.head()
```

```python
df_product = pd.read_csv('/kimia-farma-dashboard/raw/kf_product.csv') 
df_product.head()
```

```python
df_inventory = pd.read_csv('/kimia-farma-dashboard/raw/kf_inventory.csv')
df_inventory.head()
```

```python
print('transaction :', df_transaction.shape)
print('branch      :', df_branch.shape)
print('product     :', df_product.shape)
print('inventory   :', df_inventory.shape)
```

```python
print('transaction info:', df_transaction.info())
print('branch info     :', df_branch.info())
print('product info    :', df_product.info())
print('inventory info  :', df_inventory.info())
```

```python
print('Missing values in transaction :' + str(df_transaction.isnull().sum()))
print('\nMissing values in branch      :' + str(df_branch.isnull().sum()))
print('\nMissing values in product     :' + str(df_product.isnull().sum()))
print('\nMissing values in inventory   :' + str(df_inventory.isnull().sum()))
```

```python
print('Duplicates in transaction :', df_transaction.duplicated().sum())
print('Duplicates in branch      :', df_branch.duplicated().sum())
print('Duplicates in product     :', df_product.duplicated().sum())
print('Duplicates in inventory   :', df_inventory.duplicated().sum())
```

```python
df_transaction['date'] = pd.to_datetime(df_transaction['date'])
df_transaction['date'].head()
```

```python
df_transaction['year']       = df_transaction['date'].dt.year
df_transaction['month']      = df_transaction['date'].dt.month
df_transaction['month_name'] = df_transaction['date'].dt.strftime('%B')
df_transaction['quarter']    = df_transaction['date'].dt.quarter
df_transaction["day_of_week"] = df_transaction['date'].dt.day_name()
df_transaction[['date', 'year', 'month', 'month_name', 'quarter']].head()
```

```python
df_transaction['nett_sales']  = df_transaction['price'] * (1 - df_transaction['discount_percentage'])

df_transaction[['price', 'discount_percentage', 'nett_sales']].head()
```

```python
print(df_branch['provinsi'].unique())
```

```python
df_branch = df_branch.rename(columns={
    'kota'      : 'city',
    'provinsi'  : 'province'
})
```

```python
print('Before:', df_inventory.shape)

df_inventory = df_inventory.drop_duplicates()
print('After drop_duplicates:', df_inventory.shape)

print(df_inventory.isnull().sum())
```

```python
df_inventory = df_inventory.rename(columns={
    'opname_stock': 'stock_qty',
    'Inventory_ID': 'inventory_id'
})  
```

```python
df_transaction.to_csv('d:/kimia-farma-dashboard/final/kf_transaction_clean.csv', index=False)
df_branch.to_csv('d:/kimia-farma-dashboard/final/kf_branch_clean.csv', index=False)
df_product.to_csv('d:/kimia-farma-dashboard/final/kf_product_clean.csv', index=False)
df_inventory.to_csv('d:/kimia-farma-dashboard/final/kf_inventory_clean.csv', index=False)

print('All 4 tables saved!')
```
